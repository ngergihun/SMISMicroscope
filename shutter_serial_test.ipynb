```python
import serial
from time import sleep

port="COM6"
baudrate=9600
bytesize=8
parity=serial.PARITY_NONE
stopbits=serial.STOPBITS_ONE
timeout=10
write_timeout=0.5
debug=True

ser = serial.Serial(
                port,
                baudrate=baudrate,
                bytesize=bytesize,
                parity=parity,
                stopbits=stopbits,
                timeout=timeout,
                write_timeout=write_timeout,
            )
```

```python
command = "GP 0"
command = command.encode()
ser.write(command)
buffer_string = ser.read(ser.inWaiting()).decode()
print(buffer_string)
endchar = "\r\n"
buffer_string = ser.read_until(endchar.encode())
lines = buffer_string.strip()
print(lines)
```
